# Tuple to Series and Dataframe

Tuple is collection of different datatype values

```python
my_tuple = (1, 5, 3, 6, 2, 'a', 'c', 'b')
print(my_tuple)

# Convert my_tuple to a pandas Series
tuple_as_series = pd.Series(my_tuple)
print("\nTuple converted to Series:")
print(tuple_as_series)
```

```python
df_from_tuple = pd.DataFrame(my_tuple, columns=['Tuple_Elements'])
display(df_from_tuple)
```

```python
# Filter rows where 'Tuple_Elements' is 'a'
filtered_df = df_from_tuple[df_from_tuple['Tuple_Elements'] == 'a']
display(filtered_df)
```

```python
# Sort df_from_tuple by 'Tuple_Elements' column in ascending order
sorted_df = df_from_tuple.astype(str).sort_values(by='Tuple_Elements')
display(sorted_df)
```

```python
# Add a new column named 'New_Column' to df_from_tuple
# The length of the list must match the number of rows in the DataFrame
new_column_values = [10, 20, 30, 40, 50, 60]
df_from_tuple['New_Column'] = new_column_values

display(df_from_tuple)
```

```python
tuple_series = df_from_tuple['Tuple_Elements']
display(tuple_series)
```

# Date and Time Date DataTypes in Pandas

```python
from datetime import datetime, timedelta

print("--- Date and Time Date Types Practice ---")

# 1. Creating a Series of datetime objects
dates = pd.Series([
    datetime(2023, 1, 1),
    datetime(2023, 1, 15, 10, 30),
    datetime(2023, 2, 28, 23, 59, 59)
])
print("\n1. Series of datetime objects:")
display(dates)

# 2. Converting various data types to datetime
string_dates = pd.Series(['2024-03-01', '2024/03/02 14:00', 'March 3, 2024'])
converted_dates = pd.to_datetime(string_dates, format='mixed')
print("\n2. Converted string dates to datetime:")
display(converted_dates)

# 3. Accessing datetime properties
print("\n3. Accessing datetime properties (from converted_dates):")
print(f"Year: {converted_dates.dt.year.tolist()}")
print(f"Month: {converted_dates.dt.month.tolist()}")
print(f"Day: {converted_dates.dt.day.tolist()}")
print(f"Hour: {converted_dates.dt.hour.tolist()}")
print(f"Day of week (Monday=0): {converted_dates.dt.dayofweek.tolist()}")

# 4. Performing datetime arithmetic
future_dates = converted_dates + pd.Timedelta(days=7, hours=3)
print("\n4. Dates after adding 7 days and 3 hours:")
display(future_dates)

# 5. Calculating difference between dates
date_diff = future_dates - converted_dates
print("\n5. Difference between dates:")
display(date_diff)
```

```python
# Create a sample DataFrame with datetime objects and values
df_time_data = pd.DataFrame({
    'Date': converted_dates,
    'Value': [100, 150, 200]
})
display(df_time_data)
```

```python
# Extract the month from the 'Date' column
df_time_data['Month'] = df_time_data['Date'].dt.month
print("\nDataFrame with Month column:")
display(df_time_data)
```

```python
# Group by month and aggregate (e.g., sum the 'Value' column)
monthly_agg = df_time_data.groupby('Month')['Value'].sum().reset_index()
print("\nAggregated data by Month:")
display(monthly_agg)
```

# Loc & iloc

```python

# Re-defining car_data here to ensure it's available
cars = pd.Series(["BMW","Toyota","Honda","BYD","Mercedece"])
colors = pd.Series(["Black","White","Red","Blue","Merun"])
model = pd.Series([2000,2001,2002,2003,2004])
car_data = pd.DataFrame({"Car_type":cars,
                         "Color":colors,
                         "Model":model})

```

```python
print("\n--- Practicing with .loc and .iloc ---\n")

# Display the original DataFrame for reference
print("Original car_data DataFrame:")
display(car_data)

```

```python
print("\n--- Using .loc (label-based indexing) ---\n")

# 1. Select a single row by label
print("1. Select a single row (index label 0) using .loc:")
display(car_data.loc[0])

# 2. Select multiple rows by label
print("\n2. Select multiple rows (index labels 1 and 3) using .loc:")
display(car_data.loc[[1, 3]])

# 3. Select rows and specific columns by label
print("\n3. Select rows (index labels 0 to 2) and 'Car_type', 'Color' columns using .loc:")
display(car_data.loc[0:2, ['Car_type', 'Color']])

# 4. Select rows based on a boolean condition
print("\n4. Select cars with 'Model' > 2002 using .loc (boolean indexing):")
display(car_data.loc[car_data['Model'] > 2002])

```

```python
print("\n--- Using .iloc (integer-position-based indexing) ---\n")

# 1. Select a single row by integer position
print("1. Select a single row (position 0) using .iloc:")
display(car_data.iloc[0])

# 2. Select multiple rows by integer position
print("\n2. Select multiple rows (positions 1 and 3) using .iloc:")
display(car_data.iloc[[1, 3]])

# 3. Select rows and specific columns by integer position
print("\n3. Select rows (positions 0 to 2) and columns (positions 0 and 1) using .iloc:")
display(car_data.iloc[0:3, 0:2]) # Note: iloc slicing is exclusive of the end index

# 4. Select all rows and a specific column by integer position
print("\n4. Select all rows and 'Model' column (position 2) using .iloc:")
display(car_data.iloc[:, 2])
```
