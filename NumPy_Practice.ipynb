```python
#Anonymous Function
squre = lambda x: x ** 2
print(squre(5))

#Common use with map, fillter
numbers = [1,3,4,5,6,7]
squred = list(map(lambda x: x ** 2, numbers))
print(squred)
evens = list(filter(lambda x: x % 2 == 0, numbers))
evens
```

```python
numbers = [1, 3, 4, 5, 6, 7]

# List comprehension for squaring (replacing map)
squared_comp = [x ** 2 for x in numbers]
print(f"Squared: {squared_comp}")

# List comprehension for filtering evens (replacing filter)
evens_comp = [x for x in numbers if x % 2 == 0]
print(f"Evens: {evens_comp}")

# Combining both: Square only the even numbers
even_squares = [x ** 2 for x in numbers if x % 2 == 0]
print(f"Squares of evens: {even_squares}")
```

```python
def even_ssqq(numbers):
  return [x**2 for x in numbers if x % 2==0]

result = even_ssqq([1,2,3,4,5,6])
print(result)
```

### Rewriting List Comprehension as a For-Loop
The function below performs the exact same operation as `even_ssqq` but uses a standard loop structure.

```python
def even_ssqq_loop(numbers):
    new_list = []
    for x in numbers:
        if x % 2 == 0:
            new_list.append(x**2)
    return new_list

# Test the loop version
result_loop = even_ssqq_loop([1, 2, 3, 4, 5, 6])
print(f"Result using for-loop: {result_loop}")
```

```python
# Lazy evaluation - doesn't store entire list
sum_of_squares = sum(x**2 for x in range(1000000))

# Generator function
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
```

### Using the Fibonacci Generator
Since `fibonacci(n)` is a generator, we can iterate over it directly using a `for` loop to retrieve values one at a time.

```python
# Generate and print the first 10 Fibonacci numbers
print("First 10 Fibonacci numbers:")
for num in fibonacci(10):
    print(num)
```

You can specify the `axis` parameter to sort along rows or columns.

*   `axis=0`: Sort along columns (each column is sorted independently).
*   `axis=1`: Sort along rows (each row is sorted independently).

```python
print("Original random_array:\n", random_array)

# Sort along axis 0 (columns)
sorted_axis0 = np.sort(random_array, axis=0)
print("\nSorted along axis=0 (columns):\n", sorted_axis0)

# Sort along axis 1 (rows)
sorted_axis1 = np.sort(random_array, axis=1)
print("\nSorted along axis=1 (rows):\n", sorted_axis1)
```

```python
print("Original random_array:\n", random_array)

# Sort in descending order (sort ascending and then reverse)
sorted_descending = np.sort(random_array)[:, ::-1]
print("\nSorted in descending order (each row):\n", sorted_descending)

# If you want to sort the entire flattened array in descending order
sorted_descending_flattened = np.sort(random_array.flatten())[::-1]
print("\nSorted in descending order (flattened array):\n", sorted_descending_flattened)
```

```python
import matplotlib.pyplot as plt

plt.imshow(panda)
plt.axis('off') # Hide axes for a cleaner image display
plt.show()
```

```python
# Convert the color image to greyscale
# A common formula for luminance (perceived brightness) is 0.2989*R + 0.5870*G + 0.1140*B
greyscale_panda = np.dot(panda[...,:3], [0.2989, 0.5870, 0.1140])

# Display the greyscale image
plt.figure(figsize=(10, 8)) # Optional: set figure size
plt.imshow(greyscale_panda, cmap='gray') # Use 'gray' colormap for greyscale
plt.title('Greyscale Panda Image')
plt.axis('off')
plt.show()

print(f"Shape of original color image: {panda.shape}")
print(f"Shape of greyscale image: {greyscale_panda.shape}")
```

```python
print(f"Original image shape: {panda.shape}")

# Define the region of interest (e.g., crop the center of the image)
# Let's say we want to crop from row 100 to 700 and column 300 to 1000
row_start, row_end = 100, 700
col_start, col_end = 300, 1000

cropped_panda = panda[row_start:row_end, col_start:col_end]

# Display the cropped image
plt.figure(figsize=(8, 6)) # Adjust figure size as needed
plt.imshow(cropped_panda)
plt.title(f'Cropped Panda Image (Rows {row_start}-{row_end}, Cols {col_start}-{col_end})')
plt.axis('off')
plt.show()

print(f"Cropped image shape: {cropped_panda.shape}")
```
